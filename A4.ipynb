## Gender and Racial Disparities in Heart Disease Outcomes

 **Motivation & problem** 


Gender bias has been prevalent in healthcare systems since the beginning. Bias influences how symptoms are recognized, evaluated, and treated. Research has increasingly explored how gender and racial disparities affect the quality of healthcare individuals receive. I have personally experienced gender bias in healthcare, particularly as someone with a uterus, which has made me interested in understanding how these disparities affect patients' experiences and health outcomes.
For this project, I want to investigate gender and racial disparities in heart disease prevalence and mortality rates. I hope to identify patterns across demographic groups and explore whether certain populations experience disproportionately different health outcomes. I am particularly interested in understanding how gender and racial disparities overlap and what these differences might suggest about inequalities in healthcare.

**Data** 

I plan to use two publicly available datasets from the Centers for Disease Control and Prevention (CDC) and the National Center for Health Statistics (NCHS).

- CDC Chronic Disease Indicators: https://www.cdc.gov/CDI/

This dataset provides national and state-level health indicators, including heart disease-related statistics broken down by sex, race and ethnicity, and age where available. I will use this data to compare heart disease prevalence and related health indicators across demographic groups.

- NCHS Heart Disease Death Rates: https://catalog.data.gov/dataset/dqs-death-rates-for-heart-disease-by-sex-race-hispanic-origin-and-age-united-states-from-c


This dataset contains heart disease mortality rates broken down by sex, race, Hispanic origin, and age. I will use it to investigate differences in mortality rates and explore how gender and racial disparities intersect.
Both datasets are publicly accessible government data. The NCHS dataset identifies its license through the U.S. government works policy. I will verify the applicable usage terms for both sources before analysis.


Ethical considerations include recognizing limitations in demographic representation, distinguishing sex from gender identity, and avoiding assumptions that differences in health outcomes directly establish discrimination or misdiagnosis.

**Unknowns & dependencies**


Some unknowns include whether the datasets provide enough consistent demographic information to support intersectional analysis and whether the available indicators can be compared across different years. I will also need to account for differences in how prevalence and mortality are measured.
